## **Imports**

```python
import os
import plotly.express as px
import pandas as pd
```

## **Carregar datasets**

```python
dataset_dir = "../01_dades/processat/"

df_sem_prom = pd.read_csv(os.path.join(dataset_dir, "homicidio_doloso_semestral_promedio_diario.csv")).set_index("Año")
df_sem_abs = pd.read_csv(os.path.join(dataset_dir, "homicidio_doloso_semestral_frecuencia_absoluta.csv")).set_index("Año")
df_jul_prom = pd.read_csv(os.path.join(dataset_dir, "homicidio_doloso_Julio_promedio_diario.csv")).set_index("Año")
df_jul_abs = pd.read_csv(os.path.join(dataset_dir, "homicidio_doloso_Julio_frecuencia_absoluta.csv")).set_index("Año")
```

## **1r Semestre**

### **Promig diari**

```python
sem_prom = px.bar(
     df_sem_prom.mean(axis=1).rename("Víctimas homicidio doloso"), 
    title = "Promedio diario de homicidio doloso a nivel nacional (2015-2025)",
    subtitle = "Enero a julio de cada año",
    labels = { 
        "value" : "Promedio diario"
    }
)
sem_prom.update_traces(showlegend=False)
sem_prom.show()
```

### **Freqüència absoluta**

```python
sem_abs = px.bar(
     df_sem_abs.sum(axis=1).rename("Víctimas homicidio doloso"), 
    title = "Conteo de homicidio doloso a nivel nacional (2015-2025)",
    subtitle = "Enero a julio de cada año",
    labels = { 
        "value" : "Valor absoluto"
    }
)
sem_abs.update_traces(showlegend=False)
sem_abs.show()
```

## **Juliol**

### **Promig diari**

```python
jul_prom = px.bar(
    df_jul_prom, 
    title = "Promedio diario de homicidio doloso a nivel nacional (2015-2025)",
    subtitle = "Julio de cada año",
    labels = { 
        "value" : "Promedio diario"
    }
)
jul_prom.update_traces(showlegend=False)
jul_prom.show()
```

### **Freqüència absoluta**

```python
jul_abs = px.bar(
    df_jul_abs, 
    title = "Conteo de homicidio doloso a nivel nacional (2015-2025)",
    subtitle = "Julio de cada año",
    labels = { 
        "value" : "Valor absoluto"
    }
)
jul_abs.update_traces(showlegend=False)
jul_abs.show()
```
