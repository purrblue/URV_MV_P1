```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Carregar dataset
df = pd.read_csv("IDVFC_NM_dic25.csv", encoding="latin-1")
df_hd = df[(df["Subtipo de delito"] == "Homicidio doloso")]
```

```python
# Eliminar columnes juny-desembre
df_hd.drop(columns = ["Julio", "Agosto", "Septiembre", "Octubre", "Noviembre", "Diciembre"], inplace=True)

# Eliminar columnes d'informació addicional
df_hd.drop(columns = ["Clave_Ent", "Entidad", "Bien jurídico afectado", "Tipo de delito", "Subtipo de delito", "Modalidad", "Sexo", "Rango de edad"], inplace=True)
```

```python
# Agrupar per any
df_hd_yearly = df_hd.groupby("Año").sum()
df_hd_yearly.sum(axis='columns').plot.bar()
```

```python
days_of_month = {
    "Enero": 31,
    "Febrero": 28,
    "Marzo": 31,
    "Abril": 30,
    "Mayo": 31,
    "Junio": 30
}
df_hd_yearly_mean = df_hd_yearly

for month, days in days_of_month.items():
    df_hd_yearly_mean[month] = df_hd_yearly[month] / days

df_hd_yearly_mean.head()
```

```python
df_hd_yearly_mean.mean(axis='columns').plot.bar()
```
