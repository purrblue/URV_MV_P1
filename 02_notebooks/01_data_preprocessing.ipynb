## **Imports**

```python
import os
import pandas as pd
```

## **Variables generals**

```python
dies_mes = {
    "Enero": 31,
    "Febrero": 28,
    "Marzo": 31,
    "Abril": 30,
    "Mayo": 31,
    "Junio": 30,
    "Julio": 31,
    "Agosto": 31,
    "Septiembre": 30,
    "Octubre": 31, 
    "Noviembre": 30,
    "Diciembre": 31
}
```

## **Homicidio doloso**

### **Carregar dataset**

```python
dataset_dir = "../01_dades/brut/"
save_dir = "../01_dades/processat/"

dataset_nom = "IDVFC_NM_dic25.csv"
```

```python
# Carregar dataset
df_path = os.path.join(dataset_dir, dataset_nom)
df = pd.read_csv(df_path, encoding="latin-1")

# Filtrar per tipus d'homicidi
df_hd = df[(df["Subtipo de delito"] == "Homicidio doloso")]

# Eliminar columnes d'informació addicional
df_hd.drop(columns = ["Clave_Ent", "Entidad", "Bien jurídico afectado", "Tipo de delito", "Subtipo de delito", "Modalidad", "Sexo", "Rango de edad"], inplace=True)
```

### **Filtre semestral**

```python
primer_semestre = ["Enero", "Febrero", "Marzo", "Abril", "Mayo", "Junio", "Julio"]

# Eliminar columnes juny-desembre
df_hd_sem = df_hd[["Año"] + primer_semestre]
```

#### **Frecuencia absoluta**

```python
# Agrupar per any
df_hd_sem = df_hd_sem.groupby("Año").sum()
df_hd_sem.head()
```

```python
# Guardar conteig absolut per any i mes
df_hd_sem.to_csv(os.path.join(save_dir, "homicidio_doloso_semestral_frecuencia_absoluta.csv"))
```

#### **Promig diari**

```python
# Calcular promig diari
df_hd_sem_prom = df_hd_sem.copy()

for mes in primer_semestre:
    df_hd_sem_prom[mes] = df_hd_sem[mes] / dies_mes[mes]

df_hd_sem_prom.head()
```

```python
# Guardar promig diari per any i mes
df_hd_sem_prom.to_csv(os.path.join(save_dir, "homicidio_doloso_semestral_promedio_diario.csv"))
```

### **Filtre mensual (Juliol)**

```python
mes = "Julio"

# Eliminar totes les columnes menys juliol
df_hd_mes = df_hd[["Año", mes]]
```

#### **Frecuencia absoluta**

```python
# Agrupar per any
df_hd_mes = df_hd_mes.groupby("Año").sum()
df_hd_mes.head()
```

```python
# Guardar conteig absolut per any i mes
df_hd_mes.to_csv(os.path.join(save_dir, f"homicidio_doloso_{mes}_frecuencia_absoluta.csv"))
```

#### **Promig diari**

```python
# Calcular promig diari
df_hd_sem_prom = df_hd_mes.copy()
df_hd_sem_prom[mes] = df_hd_mes[mes] / dies_mes[mes]
df_hd_sem_prom.head()
```

```python
# Guardar promig diari per any i mes
df_hd_sem_prom.to_csv(os.path.join(save_dir, f"homicidio_doloso_{mes}_promedio_diario.csv")) 
```

## **Desapareguts i no localitzats**

### **Carregar datasets**

```python
desapareguts_total = []
anys = range(2015, 2026)
```

```python
for any in anys:
    dataset_nom_rnpdno = f"rnpdno_1sem_{any}.csv"
    df_path = os.path.join(dataset_dir, dataset_nom_rnpdno)
    
    # Carregar dataset 
    df_desap_anual = pd.read_csv(df_path, sep=";", encoding="latin-1")
    
    # Calcular el total de desapareguts
    total_any = df_desap_anual[["Hombre", "Mujer", "Indeterminado"]].sum(axis=None)
    
    # Afegir les dades al registre
    desapareguts_total.append({"Año": any, "Desaparecidos": total_any})
```

```python
# Crear dataset final
df_desap_total = pd.DataFrame(desapareguts_total)
df_desap_total.set_index("Año", inplace=True)
df_desap_total.head()
```

```python
# Guardar dataset desapareguts
df_desap_total.to_csv(os.path.join(save_dir, f"desaparecidos_semestral.csv"))
```
